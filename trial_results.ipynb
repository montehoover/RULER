```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import re
```

```python
baseline_dir = 'baseline/gradientai/Llama-3-8B-Instruct-262k/synthetic/8192/pred'
target_file = 'summary.csv'
baseline_2048 = pd.DataFrame()
baseline_16 = pd.DataFrame()
for subdir in os.listdir(baseline_dir):
    subdir_path = os.path.join(baseline_dir, subdir)

    # Check if it's a directory
    if os.path.isdir(subdir_path):
        # Check if the target file exists in this directory
        target_path = os.path.join(subdir_path, target_file)
        if os.path.isfile(target_path):
            print(f"Found {target_file} in {subdir_path}")
            if "2048" in subdir:
                baseline_2048 = pd.read_csv(target_path)
            else:
                baseline_16 = pd.read_csv(target_path)
```

```python
display(baseline_16)
display(baseline_2048)
```

```python
layer_dir = 'trial_outputs/1/gradientai/Llama-3-8B-Instruct-262k/synthetic/8192'

subdirs = [d for d in os.listdir(layer_dir) if os.path.isdir(os.path.join(layer_dir, d))]
sorted_subdirs = sorted(subdirs, key=lambda d: int(re.search(r'\d+$', d).group()) if re.search(r'\d+$', d) else -1)

# Initialize a list to store DataFrames
layer_dfs = []

# Process each directory in sorted order
for subdir in sorted_subdirs:
    print(f'Adding layer from: {subdir}')
    
    # Construct paths
    subdir_path = os.path.join(layer_dir, subdir)
    subdir_path = os.path.join(subdir_path, "pred/topk_2048_16")
    target_path = os.path.join(subdir_path, target_file)
    
    print(f'Target path: {target_path}')
    
    # Check if the target file exists
    if os.path.isfile(target_path):
        # Read the CSV file into a DataFrame
        layer_df = pd.read_csv(target_path)
        layer_dfs.append(layer_df)
        
        # Display the DataFrame
        print(f'DISPLAYING NOW')
        display(layer_df)
        print('\n')
```

```python
all_layers = []

for i, df in enumerate(layer_dfs):
    # Drop the row where the first column is "Nulls"
    df = df[df.iloc[:, 0] != "Nulls"]
    
    # Extract "Tasks" and "Score" rows for this layer
    tasks = df.iloc[0, 1:]  # Skip the first column (Tasks label)
    scores = df.iloc[1, 1:]  # Skip the first column (Score label)
    
    # Create a DataFrame for this layer with tasks as index
    layer_df = pd.DataFrame(scores).transpose()
    layer_df.columns = tasks  # Set tasks as column headers
    layer_df.index = [i]  # Set the layer as the index
    
    # Append this DataFrame to the list
    all_layers.append(layer_df)

# Concatenate all layers into a single DataFrame
merged_df = pd.concat(all_layers, axis=0)

# Reset index (optional) or keep layers as the index
merged_df.index.name = 'Layer'

# Display the merged DataFrame
display(merged_df)
```

```python
baseline_2048_cleaned = baseline_2048[baseline_2048.iloc[:, 0] != "Nulls"]
baseline_16_cleaned = baseline_16[baseline_16.iloc[:, 0] != "Nulls"]

# Extract "Tasks" and "Score" rows for baseline_2048
tasks_2048 = baseline_2048_cleaned.iloc[0, 1:]  # Skip the first column (Tasks label)
scores_2048 = baseline_2048_cleaned.iloc[1, 1:]  # Skip the first column (Score label)

# Create DataFrame for baseline_2048
baseline_2048_df = pd.DataFrame(scores_2048).transpose()
baseline_2048_df.columns = tasks_2048  # Set tasks as column headers
baseline_2048_df.index = [2048] * len(baseline_2048_df)  # Set the baseline as the index

# Extract "Tasks" and "Score" rows for baseline_16
tasks_16 = baseline_16_cleaned.iloc[0, 1:]  # Skip the first column (Tasks label)
scores_16 = baseline_16_cleaned.iloc[1, 1:]  # Skip the first column (Score label)

# Create DataFrame for baseline_16
baseline_16_df = pd.DataFrame(scores_16).transpose()
baseline_16_df.columns = tasks_16  # Set tasks as column headers
baseline_16_df.index = [16] * len(baseline_16_df)  # Set the baseline as the index

# Concatenate the two DataFrames
baseline_df = pd.concat([baseline_2048_df, baseline_16_df], axis=0)
baseline_df.index.name = 'baseline'
# Display the merged DataFrame
display(baseline_df)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Ensure all columns in merged_df and baseline_df are numeric
merged_df = merged_df.apply(pd.to_numeric, errors='coerce')
baseline_df = baseline_df.apply(pd.to_numeric, errors='coerce')

# Create a figure with subplots for each task
num_tasks = len(merged_df.columns)  # Number of tasks
fig, axes = plt.subplots(num_tasks, 1, figsize=(12, 6 * num_tasks))

# If there is only one task, axes is not a list, so we need to handle it as a special case
if num_tasks == 1:
    axes = [axes]

# Define a color map for layers (0-31) and baselines (2048, 16)
layer_colors = plt.cm.viridis(np.linspace(0, 1, 32))  # 32 distinct colors for layers (0-31)
baseline_colors = {'2048': 'tab:blue', '16': 'tab:orange'}  # Use distinct colors for baselines

# Calculate the global min and max for y-axis scaling (based on both merged_df and baseline_df)
all_scores = pd.concat([merged_df, baseline_df])
y_min, y_max = all_scores.min().min(), all_scores.max().max()

# Loop over each task to plot on its own subplot
for i, task in enumerate(merged_df.columns):
    ax = axes[i]
    
    # Spread the layers across the x-axis (from 0 to 31)
    x_layer_values = np.linspace(0, 31, 32)  # 32 evenly spaced values for the layers
    
    # Plot data from merged_df (layers 0-31)
    for idx, layer in enumerate(merged_df.index):
        ax.plot(x_layer_values[idx], merged_df[task][layer], 'o', color=layer_colors[idx], label=f'Layer {layer}' if idx == 0 else "")
        
        # Annotate each point with its respective layer index (0-31)
        ax.text(x_layer_values[idx], merged_df[task][layer], f'{layer}', fontsize=9, ha='right', color=layer_colors[idx])  # Layer index as label

    # Plot data from baseline_df (baselines 2048 vs 16)
    for baseline in baseline_df.index:
        ax.plot(-1, baseline_df[task][baseline], 'x', color=baseline_colors[str(baseline)], label=f'Baseline {baseline}' if baseline == 2048 else "")
        
        # Annotate each point with its respective baseline value (2048 or 16)
        ax.text(-1, baseline_df[task][baseline], f'{baseline}', fontsize=9, ha='left', color=baseline_colors[str(baseline)])  # Baseline value as label

    # Set labels and title for each subplot
    ax.set_xlabel('Layer / Baseline')
    ax.set_ylabel('Score')
    ax.set_title(f'Scores for {task}')

    # Add a legend for each subplot
    ax.legend(title="Source", loc='upper left')

    # Set x-axis limits to give space for both layers and baseline at x=-1
    ax.set_xlim(-1.5, 31.5)

    # Set the y-axis limits to ensure numerical order
    ax.set_ylim(float(y_min) - 5, float(y_max) + 5)  # Add a small margin for better visualization

# Adjust layout to prevent overlap
plt.tight_layout()

# Display the plot
plt.show()
```
